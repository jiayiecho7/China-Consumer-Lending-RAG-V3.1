## Method and limitations

The supplied PDF pages were rendered and visually inspected for headings, issuing agencies, dates, document numbers, source footers, and relevant clauses. Their extractable text largely consists of converter watermarks rather than the underlying Chinese content. D07 is clipped at the right edge. The DOC contains recoverable Chinese text; all 70 D02 article bodies were checked against it. No claim is made that all 199 chunks were independently matched verbatim.

An online cross-check succeeded for the Shanghai government D02 mirror. Requests to several gov.cn and NFRA pages returned HTTP 403, so those checks rely on the supplied printouts and are not described as successful live website verification. Printouts establish their displayed provenance, not cryptographic authenticity or absence of later amendments.
