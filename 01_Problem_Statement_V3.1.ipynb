## Problem and motivation

Consumer-lending compliance questions often require analysts to locate and interpret provisions scattered across multiple regulatory instruments. Relevant rules may depend on the loan purpose, institution, distribution channel, loan term, or charging arrangement. Manual search can be time-consuming, while a plausible answer can still omit an important condition or rely on an inappropriate provision. A useful assistant should make the underlying evidence easy to inspect, rather than merely produce a fluent answer.

This project focuses on consumer lending in mainland China. Loan categories, institutional responsibilities, and regulatory requirements differ across countries, so a single jurisdiction provides a coherent scope for retrieval and evaluation. I am from China and previously worked in a finance-related role. This background motivated my choice and gives me familiarity with the domain, while individual provisions still require verification. Original Chinese regulatory quotations are retained to preserve their conditions and exceptions; the project documentation explains the approach in English.

## Intended user and use case

The intended user is a compliance analyst reviewing a practical consumer-lending question. The analyst enters a question, inspects retrieved provisions and their sources, and reviews an evidence-supported interpretation. The assistant supports research and review; the analyst remains responsible for the final compliance decision.

**Example:** For a proposed consumer-loan term, identify the applicable term provision, quote the relevant evidence, explain its implications, and state any uncertainty about special conditions.

## Project objective

Build and evaluate a bounded retrieval-augmented generation workflow that retrieves relevant Chinese lending provisions and produces explanations with traceable source identifiers. The central experimental question is:

> On the same fifty provision-labeled questions, how does semantic retrieval compare with a plain keyword baseline, and how does hybrid retrieval with evidence selection affect top-five provision coverage?

The project also examines whether generated explanations preserve regulatory conditions, whether quotations and citations are reliable, and what model usage costs are recorded. Retrieval coverage and answer quality are evaluated separately.

## Scope and data

The historical experiment uses eight named regulatory instruments, 199 article or section chunks, and fifty hypothetical Chinese questions. The corpus covers personal lending, internet lending, partner-assisted lending, automobile lending, and financing-cost disclosure. The document inventory and corpus are included in `03_Data_and_Corpus_V3.1.ipynb`. Nine supplied source files represent eight instruments because one instrument has two copies.

This is a bounded project corpus, not an exhaustive collection of every Chinese lending rule. Applicability beyond the documented scope and amendments after the evaluation date require additional checks.

## Proposed and implemented approach

1. Organize regulatory text into identifiable article or section chunks with source metadata.
2. Retrieve candidates using semantic similarity and BM25 keyword search on the same text fields.
3. Combine candidates through reciprocal-rank fusion and supplement neighboring provisions.
4. Use GPT-4.1-mini to select evidence, draft an explanation, and audit its support. Semantic embeddings use text-embedding-3-small.
5. Present the explanation, original evidence, and source references for human inspection.

Gold labels are evaluation inputs and are not passed to the answer generator. The adopted release, V3.1, denotes V3 with Q36 repaired. The runnable Colab entry is `02_Main_Code_and_API_V3.1.ipynb`; its default demonstration reproduces saved results. A separate optional API cell supports a new run.

## Evaluation design and achieved results

The primary measure is Hit@5: the proportion of the fifty questions for which at least one expected provision appears among the first five results. The keyword baseline, semantic method, and hybrid selection are compared on the same questions.

| Historical method | Original Hit@5 |
|---|---|
| BM25 keyword retrieval | 39/50 (78%) |
| Semantic retrieval | 41/50 (82%) |
| Hybrid retrieval with LLM evidence selection | 49/50 (98%) |

Coverage within the final evidence set of up to ten provisions is 50/50 (100%), a separate measure. These percentages do not establish answer accuracy. The questions were used for development and tuning, so the results are development-set measurements rather than an independent held-out test.

Later source corrections and alternative gold mappings are documented separately in `09_Gold_and_Key_Clause_Verification_V3.1.ipynb`. Rescoring existing rankings does not constitute a new retrieval experiment. Six answer-review judgments were personally confirmed by LI JIAYI; the fifty-mapping review is AI-assisted and is not presented as complete independent human validation.

## Success criteria, trade-offs, and limitations

The practical success criteria are reproducible comparison against keyword search, transparent corpus and evaluation records, traceable evidence, and explicit treatment of unsupported interpretations. The instructor's requested top-five comparison is the primary evaluation requirement; no unsupported numerical target is attributed to the instructor.

Hybrid retrieval and evidence selection may improve provision coverage while increasing model usage and exposing the generator to irrelevant conditions. Original Chinese text preserves regulatory meaning but requires domain and language understanding. The workflow needs independent testing, broader human review, and maintained sources before practical reliance. Cache-based usage costs are documented separately and are not reconciled provider invoices.
